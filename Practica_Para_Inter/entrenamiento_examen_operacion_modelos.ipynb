## Bloque 13 · Pydantic: `Field` con límites

`Field(ge=X, le=Y)` exige que el número esté entre X e Y (incluidos).
`ge` = greater or equal, `le` = less or equal, `gt`/`lt` son sin incluir el
límite. `Field(min_length=1)` exige que un texto no esté vacío.

````python
from pydantic import BaseModel, Field

class Fruta(BaseModel):
    nombre: str = Field(min_length=1)
    peso_kg: float = Field(gt=0, le=50)

f = Fruta(nombre="pera", peso_kg="10")
print(f)

from pydantic import ValidationError
try:
    Fruta(nombre="pera", peso_kg="0")   # 0 no es > 0
except ValidationError as e:
    print("Rechazado:", e.errors()[0]["msg"])
````

## Bloque 16 · Pydantic: `field_validator` (el truco del TODO 2.7)

A veces un campo opcional llega del CSV como texto vacío `""`, y `""` no es
un número válido. Un `field_validator` con `mode="before"` te deja
**transformar el valor antes** de que Pydantic intente convertirlo:

```python
from pydantic import field_validator

class X(BaseModel):
    campo: float | None = None

    @field_validator("campo", mode="before")
    @classmethod
    def vacio_es_none(cls, value):
        if value == "":
            return None
        return value
```

Esto dice: "antes de validar `campo`, si el valor es `''`, conviértelo en
`None`; si no, déjalo tal cual para que Pydantic lo valide normalmente".

````python
from pydantic import BaseModel, field_validator

class Fruta(BaseModel):
    nombre: str
    peso_kg: float | None = None

    @field_validator("peso_kg", mode="before")
    @classmethod
    def vacio_es_none(cls, value):
        if value == "":
            return None
        return value

print(Fruta(nombre="pera", peso_kg="").peso_kg)     # None
print(Fruta(nombre="pera", peso_kg="2.5").peso_kg)  # 2.5
````
